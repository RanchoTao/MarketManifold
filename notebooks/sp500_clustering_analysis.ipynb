## 1. 运行完整实验流水线

脚本仅依赖 Python 标准库，因此在无法联网安装依赖的机房环境也能运行。

```python
import sys
from pathlib import Path

repo_root = Path.cwd().resolve()
if repo_root.name == 'notebooks':
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root / 'src'))

from market_clustering import run_pipeline

outputs = run_pipeline(
    input_path=repo_root / 'sp500_features.csv',
    results_dir=repo_root / 'results',
    final_k=8,
    k_values=list(range(2, 13)),
)
outputs
```

## 2. 查看 K 值评价结果

Silhouette Score 越大越好；Davies-Bouldin Index 越小越好；Calinski-Harabasz Index 越大越好；SSE/Inertia 用于肘部法观察。

```python
import csv

with (repo_root / 'results' / 'metrics.csv').open(encoding='utf-8') as f:
    metrics = list(csv.DictReader(f))

for row in metrics:
    print(row)
```

```python
with (repo_root / 'results' / 'cluster_profile_summary.csv').open(encoding='utf-8') as f:
    profiles = list(csv.DictReader(f))

for row in profiles:
    print(row['cluster'], row['profile_name'], row['asset_count'], row['representative_tickers'])
```
